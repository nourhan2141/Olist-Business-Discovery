# Sellers Profiling

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/olist_sellers_dataset.csv')
print('Rows, Cols:', df.shape)
df.dtypes
```

```python
df.head(10)
```

```python
missing = df.isnull().sum()
missing_pct = df.isnull().mean() * 100
pd.DataFrame({'Count': missing, 'Pct': missing_pct}).sort_values(by='Count', ascending=False)
```

```python
print('Full row duplicates:', df.duplicated().sum())
```

```python
print('seller_id duplicates:', df['seller_id'].duplicated().sum())
print('Row count == distinct seller_id:', len(df) == df['seller_id'].nunique())
```

```python
df.describe(include='O')
```

```python
df.describe() # numeric (if any)
```

```python
df['seller_state'].value_counts().head(10).plot(kind='bar', title='Top Seller States')
```

```python
df['seller_city'].value_counts().head(10).plot(kind='bar', title='Top Seller Cities')
```

### Takeaways
- There are 3095 unique sellers, matching the row count (no duplicates).
- The vast majority of sellers are concentrated in São Paulo (SP) state.
- There are no missing values in this dataset, and the primary key `seller_id` is clean and unique.
